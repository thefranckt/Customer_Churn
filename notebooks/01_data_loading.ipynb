## 1. Objective

The objective of this notebook is to load the Telco Customer Churn dataset and carry out a first inspection. This includes:

- Reading the CSV file into a DataFrame
- Verifying dimensions, datatypes, and structure
- Identifying possible data quality issues (e.g., nulls, types, duplicates)
- Saving a working copy in `data/processed/` for use in upcoming notebooks


2. Loading the Dataset

```python
import pandas as pd
import numpy as np
import os

# Set display options
pd.set_option('display.max_columns', None)

# Path to dataset
DATA_PATH = "../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv"

# Load data
try:
    df = pd.read_csv(DATA_PATH)
    print("✅ Dataset loaded successfully.")
except FileNotFoundError:
    print("❌ Dataset not found. Please check the path:", DATA_PATH)

# Preview
df.head()
```

3. Initial Data Inspection

```python
# Shape of the dataset
print(f"📊 Rows: {df.shape[0]}, Columns: {df.shape[1]}")

# Column data types
df.dtypes

# Null values per column
print("🕳️ Missing values per column:\n")
print(df.isnull().sum())

# Basic descriptive statistics
df.describe(include='all').T
```

4. Save Working

```python
# Save a working version in /data/processed
output_path = "../data/processed/telco_clean_base.csv"
df.to_csv(output_path, index=False)
print(f"💾 Data saved to: {output_path}")
```

- The dataset was successfully loaded with 7043 rows and 21 columns.
- We inspected the data types, basic statistics, and missing values.
- A base working version of the dataset was saved for future processing and modeling steps.
